```python
%pip install pandas
%pip install seaborn
import os
import numpy as np
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
from keras import layers, models, Input
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from PIL import Image
from glob import glob
import pandas as pd
import seaborn as sns
from datetime import datetime

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Configuration parameters
IMG_SIZE = (128, 128)  # Increased image size for better feature extraction
BATCH_SIZE = 32
EPOCHS = 100
LEARNING_RATE = 0.001
EARLY_STOPPING_PATIENCE = 10

def load_images_and_labels(folder, verbose=True):
    """Load images and extract labels from filenames with robust error handling."""
    image_paths = glob(os.path.join(folder, '*.[bB][mM][pP]'))
    
    if verbose:
        print(f"Found {len(image_paths)} images in {folder}")
    
    if not image_paths:
        raise ValueError(f"No images found in {folder}")
    
    X, user_ids, genders, fingers = [], [], [], []
    skipped_files = 0

    for path in image_paths:
        filename = os.path.basename(path)
        try:
            parts = filename.split("__", 1)
            if len(parts) != 2:
                raise ValueError(f"Invalid filename format: {filename}")
            
            user_id = parts[0]
            remaining = parts[1]

            components = remaining.split("_")
            if len(components) < 4:
                raise ValueError(f"Invalid finger format in {filename}")
            
            gender = components[0].lower()  # Standardize gender capitalization
            finger = "_".join(components[1:-1])

            # Open image, convert to grayscale, resize
            img = Image.open(path).convert('L').resize(IMG_SIZE)
            img_array = np.array(img)
            
            # Basic quality check - skip very dark or very bright images
            if np.mean(img_array) < 10 or np.mean(img_array) > 245:
                if verbose:
                    print(f"Skipping {filename}: poor contrast")
                skipped_files += 1
                continue
                
            X.append(img_array)
            user_ids.append(user_id)
            genders.append(gender)
            fingers.append(finger)

        except Exception as e:
            if verbose:
                print(f"Error loading {filename}: {e}")
            skipped_files += 1

    if not X:
        raise ValueError(f"No valid images loaded from {folder}")
    
    if verbose and skipped_files > 0:
        print(f"Skipped {skipped_files} problematic files")
    
    # Normalize to [0,1] and add channel dimension
    X = np.array(X)[..., np.newaxis] / 255.0
    return X, user_ids, genders, fingers

def data_augmentation_layer():
    """Create a data augmentation pipeline for training."""
    return keras.Sequential([
        layers.RandomRotation(0.05),
        layers.RandomTranslation(0.1, 0.1),
        layers.RandomZoom(0.1),
        layers.RandomContrast(0.2),
    ], name="data_augmentation")

def encode_labels(user_ids, genders, fingers):
    """Encode categorical labels and return encoders."""
    user_encoder = LabelEncoder()
    gender_encoder = LabelEncoder()
    finger_encoder = LabelEncoder()

    y_user = user_encoder.fit_transform(user_ids)
    y_gender = gender_encoder.fit_transform(genders)
    y_finger = finger_encoder.fit_transform(fingers)

    # Print class distributions
    print(f"\nUser distribution: {len(user_encoder.classes_)} unique users")
    print(f"Gender distribution: {np.bincount(y_gender)} ({gender_encoder.classes_})")
    print(f"Finger distribution: {np.bincount(y_finger)} ({finger_encoder.classes_})")

    return (y_user, y_gender, y_finger), (user_encoder, gender_encoder, finger_encoder)

def build_enhanced_model(num_users, num_genders, num_fingers, use_augmentation=True):
    """Build an enhanced model with residual connections and regularization."""
    inputs = Input(shape=(*IMG_SIZE, 1))
    
    # Optional data augmentation (only applied during training)
    if use_augmentation:
        x = data_augmentation_layer()(inputs)
    else:
        x = inputs
        
    # First block
    conv1 = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(x)
    bn1 = layers.BatchNormalization()(conv1)
    act1 = layers.Activation('relu')(bn1)
    pool1 = layers.MaxPooling2D((2, 2))(act1)
    
    # Second block with residual connection
    conv2a = layers.Conv2D(64, (3, 3), padding='same')(pool1)
    bn2a = layers.BatchNormalization()(conv2a)
    act2a = layers.Activation('relu')(bn2a)
    
    conv2b = layers.Conv2D(64, (3, 3), padding='same')(act2a)
    bn2b = layers.BatchNormalization()(conv2b)
    
    # Residual connection
    res2 = layers.Conv2D(64, (1, 1), padding='same')(pool1)  # Adjust dimensions
    add2 = layers.Add()([bn2b, res2])
    act2b = layers.Activation('relu')(add2)
    pool2 = layers.MaxPooling2D((2, 2))(act2b)
    
    # Third block with residual connection
    conv3a = layers.Conv2D(128, (3, 3), padding='same')(pool2)
    bn3a = layers.BatchNormalization()(conv3a)
    act3a = layers.Activation('relu')(bn3a)
    
    conv3b = layers.Conv2D(128, (3, 3), padding='same')(act3a)
    bn3b = layers.BatchNormalization()(conv3b)
    
    # Residual connection
    res3 = layers.Conv2D(128, (1, 1), padding='same')(pool2)  # Adjust dimensions
    add3 = layers.Add()([bn3b, res3])
    act3b = layers.Activation('relu')(add3)
    pool3 = layers.MaxPooling2D((2, 2))(act3b)
    
    # Fourth block
    conv4 = layers.Conv2D(256, (3, 3), activation='relu', padding='same')(pool3)
    bn4 = layers.BatchNormalization()(conv4)
    act4 = layers.Activation('relu')(bn4)
    pool4 = layers.MaxPooling2D((2, 2))(act4)
    
    # Global pooling and flatten
    gap = layers.GlobalAveragePooling2D()(pool4)
    flatten = layers.Flatten()(pool4)
    
    # Concatenate global pooling with flattened features
    concat = layers.Concatenate()([gap, flatten])
    
    # Dense layers
    dense1 = layers.Dense(512, activation='relu')(concat)
    bn_dense1 = layers.BatchNormalization()(dense1)
    dropout1 = layers.Dropout(0.5)(bn_dense1)
    
    dense2 = layers.Dense(256, activation='relu')(dropout1)
    bn_dense2 = layers.BatchNormalization()(dense2)
    dropout2 = layers.Dropout(0.3)(bn_dense2)
    
    # Task-specific layers
    # User identification - the most complex task
    user_dense = layers.Dense(256, activation='relu', name='user_dense')(dropout2)
    user_output = layers.Dense(num_users, activation='softmax', name='user_output')(user_dense)
    
    # Gender classification - simpler task
    gender_dense = layers.Dense(128, activation='relu', name='gender_dense')(dropout2)
    gender_output = layers.Dense(num_genders, activation='softmax', name='gender_output')(gender_dense)
    
    # Finger classification
    finger_dense = layers.Dense(128, activation='relu', name='finger_dense')(dropout2)
    finger_output = layers.Dense(num_fingers, activation='softmax', name='finger_output')(finger_dense)

    # Create model
    model = models.Model(inputs=inputs, outputs=[user_output, gender_output, finger_output])
    
    # Compile with different weights for each task
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=LEARNING_RATE),
        loss={
            'user_output': 'sparse_categorical_crossentropy',
            'gender_output': 'sparse_categorical_crossentropy',
            'finger_output': 'sparse_categorical_crossentropy'
        },
        loss_weights={
            'user_output': 1.0,  # Primary task
            'gender_output': 0.5,  # Secondary tasks
            'finger_output': 0.5
        },
        metrics={
            'user_output': ['accuracy', keras.metrics.SparseTopKCategoricalAccuracy(k=5, name='top5_acc')],
            'gender_output': ['accuracy'],
            'finger_output': ['accuracy']
        }
    )
    return model

def train_valid_test_split(X, y_user, y_gender, y_finger, valid_size=0.15, test_size=0.15):
    """Split data into train, validation, and test sets."""
    # First split: separate test set
    X_temp, X_test, y_user_temp, y_user_test, y_gender_temp, y_gender_test, y_finger_temp, y_finger_test = train_test_split(
        X, y_user, y_gender, y_finger, test_size=test_size, random_state=42, stratify=y_user
    )
    
    # Second split: separate validation set from remaining data
    valid_adjusted_size = valid_size / (1 - test_size)
    X_train, X_valid, y_user_train, y_user_valid, y_gender_train, y_gender_valid, y_finger_train, y_finger_valid = train_test_split(
        X_temp, y_user_temp, y_gender_temp, y_finger_temp, test_size=valid_adjusted_size, random_state=42, stratify=y_user_temp
    )
    
    print(f"Train set: {X_train.shape[0]} samples")
    print(f"Validation set: {X_valid.shape[0]} samples")
    print(f"Test set: {X_test.shape[0]} samples")
    
    return (X_train, y_user_train, y_gender_train, y_finger_train), \
           (X_valid, y_user_valid, y_gender_valid, y_finger_valid), \
           (X_test, y_user_test, y_gender_test, y_finger_test)

def plot_confusion_matrix(y_true, y_pred, classes, title, figsize=(10, 8)):
    """Plot a confusion matrix."""
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=figsize)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
    plt.title(title)
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    return plt

def visualize_model_performance(history):
    """Visualize training history with detailed plots."""
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    
    # Plot accuracy
    axes[0, 0].plot(history.history['user_output_accuracy'], label='Train')
    axes[0, 0].plot(history.history['val_user_output_accuracy'], label='Validation')
    axes[0, 0].set_title('User ID Accuracy')
    axes[0, 0].set_xlabel('Epoch')
    axes[0, 0].set_ylabel('Accuracy')
    axes[0, 0].legend()
    
    axes[0, 1].plot(history.history['gender_output_accuracy'], label='Train')
    axes[0, 1].plot(history.history['val_gender_output_accuracy'], label='Validation')
    axes[0, 1].set_title('Gender Accuracy')
    axes[0, 1].set_xlabel('Epoch')
    axes[0, 1].legend()
    
    axes[0, 2].plot(history.history['finger_output_accuracy'], label='Train')
    axes[0, 2].plot(history.history['val_finger_output_accuracy'], label='Validation')
    axes[0, 2].set_title('Finger Accuracy')
    axes[0, 2].set_xlabel('Epoch')
    axes[0, 2].legend()
    
    # Plot loss
    axes[1, 0].plot(history.history['user_output_loss'], label='Train')
    axes[1, 0].plot(history.history['val_user_output_loss'], label='Validation')
    axes[1, 0].set_title('User ID Loss')
    axes[1, 0].set_xlabel('Epoch')
    axes[1, 0].set_ylabel('Loss')
    axes[1, 0].legend()
    
    axes[1, 1].plot(history.history['gender_output_loss'], label='Train')
    axes[1, 1].plot(history.history['val_gender_output_loss'], label='Validation')
    axes[1, 1].set_title('Gender Loss')
    axes[1, 1].set_xlabel('Epoch')
    axes[1, 1].legend()
    
    axes[1, 2].plot(history.history['finger_output_loss'], label='Train')
    axes[1, 2].plot(history.history['val_finger_output_loss'], label='Validation')
    axes[1, 2].set_title('Finger Loss')
    axes[1, 2].set_xlabel('Epoch')
    axes[1, 2].legend()
    
    plt.tight_layout()
    return fig

def visualize_sample_predictions(model, X_samples, y_true_user, y_true_gender, y_true_finger, encoders, n_samples=5):
    """Visualize sample predictions with their true and predicted labels."""
    user_encoder, gender_encoder, finger_encoder = encoders
    
    # Get predictions
    preds = model.predict(X_samples[:n_samples])
    pred_users = np.argmax(preds[0], axis=1)
    pred_genders = np.argmax(preds[1], axis=1)
    pred_fingers = np.argmax(preds[2], axis=1)
    
    # Create figure
    fig, axes = plt.subplots(n_samples, 1, figsize=(12, 4*n_samples))
    
    for i in range(n_samples):
        ax = axes[i] if n_samples > 1 else axes
        
        # Display image
        ax.imshow(X_samples[i, :, :, 0], cmap='gray')
        
        # Add predictions as text
        true_user = user_encoder.inverse_transform([y_true_user[i]])[0]
        pred_user = user_encoder.inverse_transform([pred_users[i]])[0]
        true_gender = gender_encoder.inverse_transform([y_true_gender[i]])[0]
        pred_gender = gender_encoder.inverse_transform([pred_genders[i]])[0]
        true_finger = finger_encoder.inverse_transform([y_true_finger[i]])[0]
        pred_finger = finger_encoder.inverse_transform([pred_fingers[i]])[0]
        
        title = f"User: {true_user} (Pred: {pred_user})\n"
        title += f"Gender: {true_gender} (Pred: {pred_gender})\n"
        title += f"Finger: {true_finger} (Pred: {pred_finger})"
        
        # Color the title based on accuracy
        if true_user == pred_user and true_gender == pred_gender and true_finger == pred_finger:
            ax.set_title(title, color='green')
        else:
            ax.set_title(title, color='red')
            
        ax.axis('off')
    
    plt.tight_layout()
    return fig

def main():
    """Main function to run the fingerprint recognition system."""
    print("🔍 Enhanced Fingerprint Recognition System")
    print("=" * 50)
    
    # Set folder paths
    train_folder = r'/path/to/preprocessed-train-images'
    test_folder = r'/path/to/real-images'

    try:
        # Load and preprocess data
        print("\n=== Loading Training Data ===")
        X_train_data, uid_train_data, gen_train_data, fin_train_data = load_images_and_labels(train_folder)
        
        print("\n=== Loading Test Data ===")
        X_test_external, uid_test_external, gen_test_external, fin_test_external = load_images_and_labels(test_folder)
        
        # Encode labels
        print("\n=== Encoding Labels ===")
        (y_user_train_data, y_gender_train_data, y_finger_train_data), encoders = encode_labels(
            uid_train_data, gen_train_data, fin_train_data
        )
        
        # Split training data into train/validation/test
        print("\n=== Splitting Data ===")
        (X_train, y_user_train, y_gender_train, y_finger_train), \
        (X_valid, y_user_valid, y_gender_valid, y_finger_valid), \
        (X_test, y_user_test, y_gender_test, y_finger_test) = train_valid_test_split(
            X_train_data, y_user_train_data, y_gender_train_data, y_finger_train_data
        )
        
        # Validate external test labels
        print("\n=== Validating External Test Labels ===")
        unknown_users = set(uid_test_external) - set(encoders[0].classes_)
        unknown_genders = set(gen_test_external) - set(encoders[1].classes_)
        unknown_fingers = set(fin_test_external) - set(encoders[2].classes_)

        if unknown_users or unknown_genders or unknown_fingers:
            print("Warning: Found unknown classes in external test set:")
            if unknown_users: print(f"Unknown users: {unknown_users}")
            if unknown_genders: print(f"Unknown genders: {unknown_genders}")
            if unknown_fingers: print(f"Unknown fingers: {unknown_fingers}")
            print("Using internal test set for final evaluation.")
            use_external_test = False
        else:
            # Encode external test labels
            print("\n=== Encoding External Test Labels ===")
            y_user_external = encoders[0].transform(uid_test_external)
            y_gender_external = encoders[1].transform(gen_test_external)
            y_finger_external = encoders[2].transform(fin_test_external)
            use_external_test = True
            
        # Build model
        print("\n=== Building Enhanced Model ===")
        model = build_enhanced_model(
            len(encoders[0].classes_), 
            len(encoders[1].classes_), 
            len(encoders[2].classes_)
        )
        model.summary()
        
        # Callbacks
        callbacks = [
            tf.keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=EARLY_STOPPING_PATIENCE,
                restore_best_weights=True,
                verbose=1
            ),
            tf.keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=5,
                min_lr=1e-6,
                verbose=1
            ),
            tf.keras.callbacks.ModelCheckpoint(
                filepath=f"fingerprint_model_{datetime.now().strftime('%Y%m%d_%H%M')}.h5",
                monitor='val_loss',
                save_best_only=True,
                verbose=1
            )
        ]
        
        # Train model
        print("\n=== Training Enhanced Model ===")
        history = model.fit(
            X_train,
            {
                'user_output': y_user_train,
                'gender_output': y_gender_train,
                'finger_output': y_finger_train
            },
            epochs=EPOCHS,
            batch_size=BATCH_SIZE,
            validation_data=(
                X_valid,
                {
                    'user_output': y_user_valid,
                    'gender_output': y_gender_valid,
                    'finger_output': y_finger_valid
                }
            ),
            callbacks=callbacks,
            verbose=1
        )
        
        # Evaluate on test set (internal)
        print("\n=== Evaluating on Internal Test Set ===")
        internal_results = model.evaluate(
            X_test,
            {
                'user_output': y_user_test,
                'gender_output': y_gender_test,
                'finger_output': y_finger_test
            },
            verbose=1
        )
        
        # Get predictions for metrics
        internal_preds = model.predict(X_test)
        
        # Calculate detailed metrics
        user_pred_internal = np.argmax(internal_preds[0], axis=1)
        gender_pred_internal = np.argmax(internal_preds[1], axis=1)
        finger_pred_internal = np.argmax(internal_preds[2], axis=1)
        
        # Print classification reports
        print("\n=== Detailed Metrics (Internal Test Set) ===")
        print("\nUser ID Classification Report:")
        print(classification_report(y_user_test, user_pred_internal))
        
        print("\nGender Classification Report:")
        print(classification_report(y_gender_test, gender_pred_internal, 
                                  target_names=encoders[1].classes_))
        
        print("\nFinger Classification Report:")
        print(classification_report(y_finger_test, finger_pred_internal, 
                                  target_names=encoders[2].classes_))
        
        # Display confusion matrices
        print("\n=== Generating Confusion Matrices ===")
        plt_gender = plot_confusion_matrix(
            y_gender_test, gender_pred_internal, 
            encoders[1].classes_, 
            "Gender Classification Confusion Matrix"
        )
        plt_gender.savefig("gender_confusion_matrix.png")
        
        plt_finger = plot_confusion_matrix(
            y_finger_test, finger_pred_internal, 
            encoders[2].classes_, 
            "Finger Classification Confusion Matrix"
        )
        plt_finger.savefig("finger_confusion_matrix.png")
        
        # If possible, evaluate on external test set
        if use_external_test:
            print("\n=== Evaluating on External Test Set ===")
            external_results = model.evaluate(
                X_test_external,
                {
                    'user_output': y_user_external,
                    'gender_output': y_gender_external,
                    'finger_output': y_finger_external
                },
                verbose=1
            )
            
            # Get predictions for metrics
            external_preds = model.predict(X_test_external)
            
            # Calculate detailed metrics
            user_pred_external = np.argmax(external_preds[0], axis=1)
            gender_pred_external = np.argmax(external_preds[1], axis=1)
            finger_pred_external = np.argmax(external_preds[2], axis=1)
            
            # Accuracy scores for external test set
            user_acc_ext = accuracy_score(y_user_external, user_pred_external)
            gender_acc_ext = accuracy_score(y_gender_external, gender_pred_external)
            finger_acc_ext = accuracy_score(y_finger_external, finger_pred_external)
            
            print("\n📊 External Test Set Accuracy Scores:")
            print(f"User ID Accuracy: {user_acc_ext:.4f}")
            print(f"Gender Accuracy: {gender_acc_ext:.4f}")
            print(f"Finger Class Accuracy: {finger_acc_ext:.4f}")
        
        # Visualize training history
        print("\n=== Visualizing Training History ===")
        vis_fig = visualize_model_performance(history)
        vis_fig.savefig("training_history.png")
        
        # Visualize sample predictions
        print("\n=== Visualizing Sample Predictions ===")
        sample_fig = visualize_sample_predictions(
            model, X_test, y_user_test, y_gender_test, y_finger_test, encoders
        )
        sample_fig.savefig("sample_predictions.png")
        
        # Save model
        model_filename = f"fingerprint_model_final_{datetime.now().strftime('%Y%m%d_%H%M')}.h5"
        model.save(model_filename)
        print(f"\n✅ Model saved as {model_filename}")
        
        print("\n🎉 Fingerprint Recognition System Completed Successfully!")
        
    except Exception as e:
        print(f"\n❌ Error: {e}")
        import traceback
        traceback.print_exc()

if __name__ == "__main__":
    main()
```
